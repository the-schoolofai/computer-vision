## 1. What is Computer Vision?

**Computer Vision (CV)** is the field of AI that enables machines to **see, understand, and reason about visual information** - images, videos, 3D scenes, and live camera feeds - the way humans do (and often better).

### The core question CV tries to answer:
> *"Given pixels, what's actually in the scene, where is it, and what does it mean?"*

### Vision vs. Image Processing - know the difference

| Image Processing | Computer Vision |
|---|---|
| Input: image -> Output: **image** | Input: image -> Output: **understanding** |
| Blur, sharpen, resize, denoise | Detect, classify, segment, describe |
| Mostly deterministic math | Learning-based, probabilistic |
| OpenCV `cv2.GaussianBlur()` | YOLO detecting a pedestrian |

### The 4 levels of vision tasks (in order of difficulty)

1. **Classification** - *What is in this image?* -> "cat"
2. **Detection** - *What and where?* -> bounding boxes around each cat
3. **Segmentation** - *Which exact pixels?* -> pixel-perfect cat mask
4. **Understanding** - *What's happening?* -> "A cat is knocking a glass off the table"


## 2. The Complete History of Computer Vision

A concise, professional timeline - from a summer project in 1966 to the multimodal foundation models of 2026.

### Era 1 - The Optimistic Beginning (1950s - 1970s)

| Year | Milestone | Why it matters |
|---|---|---|
| **1959** | Hubel & Wiesel record cat visual cortex neurons | Discovered **edge detectors** in the brain -> blueprint for CNNs |
| **1963** | Larry Roberts' PhD thesis (MIT) | First-ever CV system: extracted 3D info from 2D blocks. Considered the **birth of CV** |
| **1966** | "The Summer Vision Project" (MIT, Marvin Minsky) | Famously assigned to undergrads for *one summer*. They thought vision was easy. It wasn't. |
| **1970s** | David Marr's computational theory of vision | Proposed the *primal sketch -> 2.5D sketch -> 3D model* hierarchy |

### Era 2 - Classical CV & Hand-crafted Features (1980s - 2000s)

| Year | Milestone | Why it matters |
|---|---|---|
| **1980** | Fukushima's **Neocognitron** | First convolutional architecture (predecessor of CNNs) |
| **1986** | Canny Edge Detector | Still used today |
| **1989** | LeCun's **LeNet** trained with backprop | First real CNN - read handwritten zip codes for the US Post Office |
| **1999** | David Lowe's **SIFT** | Scale-invariant features -> revolutionized matching, stitching, SLAM |
| **2001** | **Viola-Jones** face detector | First *real-time* face detection. Powered every digital camera for a decade |
| **2005** | **HOG** features (Dalal & Triggs) | Pedestrian detection - the bridge to modern detectors |

### Era 3 - The Deep Learning Revolution (2012 - 2017)

| Year | Milestone | Why it matters |
|---|---|---|
| **2009** | **ImageNet** dataset released (Fei-Fei Li) | 14M labeled images - fuel for the revolution |
| **2012** | **AlexNet** wins ImageNet by a huge margin | The "Big Bang" moment. CV was never the same |
| **2014** | **VGGNet**, **GoogLeNet**, **GANs** (Goodfellow) | Deeper networks + generative models |
| **2015** | **ResNet** (152 layers, skip connections) | Beat humans on ImageNet classification |
| **2015** | **U-Net**, **Faster R-CNN**, **YOLO v1** | Modern detection & segmentation born |
| **2017** | "Attention Is All You Need" (Transformers) | Indirectly reshapes CV from 2020 onward |

### Era 4 - Foundation Models & Multimodality (2020 - 2024)

| Year | Milestone | Why it matters |
|---|---|---|
| **2020** | **Vision Transformer (ViT)** | Transformers beat CNNs on image classification |
| **2021** | **CLIP** (OpenAI) | Aligned vision and language -> zero-shot classification |
| **2022** | **Stable Diffusion**, **DALL-E 2**, **Imagen** | Generative vision goes mainstream |
| **2023** | **SAM** (Segment Anything, Meta) | Foundation model for segmentation - segment *anything* with a click |
| **2023** | **GPT-4V**, **LLaVA**, **Gemini Vision** | Vision-Language Models (VLMs) become consumer tools |
| **2024** | **SAM 2**, **YOLOv10/v11**, **Sora**, **DINOv2** | Real-time video understanding, generation, self-supervised vision |

### Era 5 - Where we are now (2025 - 2026)

- Unified multimodal models - single models doing image + video + audio + text + action (Gemini 2.5, GPT-4o successors, Claude with vision)
- 3D & World Models - NeRF -> Gaussian Splatting -> real-time 3D reconstruction
- Embodied vision - robotics + CV merging (RT-2, Open X-Embodiment)
- On-device vision - Apple Vision Pro, edge YOLO, mobile diffusion
- Video-native models - generation (Sora 2, Veo) and understanding at scale

> **Takeaway:** CV evolved from *hand-engineered features* -> *learned features (CNNs)* -> *foundation models* -> *unified multimodal reasoners*. Each shift made the previous "specialist" tools obsolete.


## 3. Core Concepts Every CV Engineer Must Know

A quick-reference glossary. We'll go deep on each in later notebooks - for now, just absorb the vocabulary.

### Image Representation

- **Pixel** - smallest unit of an image; carries intensity (grayscale) or color (RGB)
- **Channel** - Grayscale = 1, RGB = 3, RGBA = 4, Hyperspectral = many
- **Resolution** - width x height (e.g., 1920 x 1080)
- **Bit depth** - usually 8-bit per channel (0-255)
- **Color spaces** - RGB, BGR (OpenCV default!), HSV, LAB, YCbCr, Grayscale

### Classical Operations

- **Convolution** - sliding a kernel over an image -> the building block of CV
- **Filtering** - blur, sharpen, edge-detect (Sobel, Laplacian, Canny)
- **Morphology** - erosion, dilation, opening, closing
- **Histograms** - distribution of pixel intensities
- **Thresholding** - binary segmentation (Otsu's method)

### Features

- **Keypoint** - a distinctive image location (corner, blob)
- **Descriptor** - a vector describing the area around a keypoint (SIFT, ORB, BRIEF)
- **Feature matching** - finding the same point across two images

### Deep Learning

- **CNN (Convolutional Neural Network)** - backbone of modern CV
- **Backbone / Neck / Head** - the 3-part anatomy of detection models
- **Transfer learning** - start from a pretrained model, fine-tune on your data
- **Data augmentation** - flips, crops, color jitter -> free training data
- **IoU (Intersection over Union)** - overlap metric for bounding boxes / masks
- **mAP (mean Average Precision)** - standard detection benchmark

### Modern Paradigms

- **Self-supervised learning** - learn from unlabeled images (DINO, MAE)
- **Zero-shot** - classify classes the model never saw labels for (CLIP)
- **Foundation model** - pretrained on huge data, adapts to many tasks (SAM, DINOv2)
- **Vision-Language Model (VLM)** - image + text in, text out (LLaVA, Gemini, Claude)

### 6.1 - Generate a sample image (so this notebook runs anywhere)

Instead of downloading from the internet, we'll create a small synthetic image. This guarantees the notebook works offline, in CI, and on any machine.


```python
# Create a synthetic 256x256 RGB image: a colored gradient with shapes
H, W = 256, 256
img = np.zeros((H, W, 3), dtype=np.uint8)

# Gradient background (R goes left->right, B goes top->bottom)
xs = np.linspace(0, 255, W, dtype=np.uint8)
ys = np.linspace(0, 255, H, dtype=np.uint8)
img[..., 0] = xs[None, :]   # Red channel
img[..., 2] = ys[:, None]   # Blue channel
img[..., 1] = 80            # Constant green

# Draw some shapes on top using OpenCV
cv2.circle(img, (180, 80), 40, (255, 255, 255), -1)        # white filled circle
cv2.rectangle(img, (40, 150), (130, 220), (0, 255, 0), -1) # green rectangle
cv2.line(img, (0, 0), (W-1, H-1), (0, 0, 0), 2)            # black diagonal

print(f"Image shape: {img.shape} (Height, Width, Channels)")
print(f"Dtype: {img.dtype}")
print(f"Min / Max: {img.min()} / {img.max()}")
print(f"Memory: {img.nbytes / 1024:.1f} KB")
```

### 6.2 - Visualize the image

> **Classic gotcha:** OpenCV reads images as **BGR**, but Matplotlib expects **RGB**.
> If colors look swapped -> you forgot to convert. Every CV engineer has been bitten by this at least once.


```python
# Show the same image 3 ways: RGB, Grayscale, and per-channel split
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

fig, axes = plt.subplots(1, 5, figsize=(16, 3.5))
titles = ["Original (RGB)", "Grayscale", "Red channel", "Green channel", "Blue channel"]
images = [img, gray, img[..., 0], img[..., 1], img[..., 2]]
cmaps = [None, "gray", "Reds", "Greens", "Blues"]

for ax, t, im, cm in zip(axes, titles, images, cmaps):
    ax.imshow(im, cmap=cm)
    ax.set_title(t, fontsize=10)
    ax.axis("off")

plt.tight_layout()
plt.show()

print("\nAn image is just a NumPy array. Everything else is math on top of that.")
```

### 6.3 - One classical operation: Canny edge detection (1986, still useful in 2026)

Edge detection finds where pixel intensity changes sharply - usually the boundaries of objects. The **Canny detector** is a 4-step classic: noise smoothing -> gradient -> non-max suppression -> hysteresis thresholding.


```python
# Apply Canny edge detection
blurred = cv2.GaussianBlur(gray, (5, 5), 1.4)   # smooth noise first
edges = cv2.Canny(blurred, threshold1=50, threshold2=150)

fig, axes = plt.subplots(1, 3, figsize=(11, 4))
axes[0].imshow(gray, cmap="gray")
axes[0].set_title("Input (grayscale)")
axes[0].axis("off")

axes[1].imshow(blurred, cmap="gray")
axes[1].set_title("After Gaussian blur")
axes[1].axis("off")

axes[2].imshow(edges, cmap="gray")
axes[2].set_title("Canny edges")
axes[2].axis("off")

plt.tight_layout()
plt.show()

print(f"Edge pixels: {(edges > 0).sum():,} out of {edges.size:,} "
      f"({100 * (edges > 0).mean():.1f}% of the image)")
```

### 6.4 - One classical operation: Image filtering (kernels in action)

Every CNN layer is just **learned** convolutional kernels. Here we manually apply hand-crafted ones - the same idea, hand-coded.


```python
# Three classic 3x3 kernels
sharpen = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
], dtype=np.float32)

emboss = np.array([
    [-2, -1, 0],
    [-1, 1, 1],
    [0, 1, 2]
], dtype=np.float32)

sobel_x = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
], dtype=np.float32)

out_sharp = cv2.filter2D(img, -1, sharpen)
out_emboss = cv2.filter2D(gray, -1, emboss)
out_sobel = cv2.filter2D(gray, cv2.CV_64F, sobel_x)
out_sobel = np.clip(np.abs(out_sobel), 0, 255).astype(np.uint8)

fig, axes = plt.subplots(1, 4, figsize=(14, 3.5))
axes[0].imshow(img)
axes[0].set_title("Original")
axes[1].imshow(out_sharp)
axes[1].set_title("Sharpened")
axes[2].imshow(out_emboss, cmap="gray")
axes[2].set_title("Emboss")
axes[3].imshow(out_sobel, cmap="gray")
axes[3].set_title("Sobel-X (vertical edges)")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()

print("A modern CNN learns thousands of kernels like these - automatically - from data.")
```

### 6.5 - Mini "from-scratch" convolution (so you really understand it)

CV deep learning is built on this single operation. Implementing it once removes the mystery forever.


```python
def convolve2d(image: np.ndarray, kernel: np.ndarray) -> np.ndarray:
    """Naive 2D convolution (educational - slow but transparent)."""
    kh, kw = kernel.shape
    pad_h, pad_w = kh // 2, kw // 2
    padded = np.pad(image, ((pad_h, pad_h), (pad_w, pad_w)), mode="edge")
    out = np.zeros_like(image, dtype=np.float32)
    for y in range(image.shape[0]):
        for x in range(image.shape[1]):
            region = padded[y:y + kh, x:x + kw]
            out[y, x] = (region * kernel).sum()
    return np.clip(out, 0, 255).astype(np.uint8)


# Compare our handmade conv vs OpenCV's optimized one on a tiny crop (for speed)
crop = gray[80:160, 80:160]
ours = convolve2d(crop, sharpen)
theirs = cv2.filter2D(crop, -1, sharpen)

fig, axes = plt.subplots(1, 3, figsize=(9, 3.2))
axes[0].imshow(crop, cmap="gray")
axes[0].set_title("Input crop")
axes[1].imshow(ours, cmap="gray")
axes[1].set_title("Our convolve2d()")
axes[2].imshow(theirs, cmap="gray")
axes[2].set_title("cv2.filter2D()")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()

print("Same result. OpenCV is just much faster (vectorized C++).")
```
